# Action detections on the Volleyball dataset

Runs the project's action model (YOLOv11m: block, receive, serve, set, spike) on the 9 annotated frames of every
validation and test clip of the Volleyball dataset (Ibrahim et al., CVPR 2016) and saves the raw detections.
Scoring is cheap and runs anywhere: `scripts/eval_actions_volleyball.py evaluate` (parameters chosen on the
validation matches, reported on the test matches).

Runs on **Kaggle** (GPU, Internet on) with inputs `sherif31/group-activity-recognition-volleyball` (data) and
`yoga0702/vball-action-model` (private: the action model). Output: `/kaggle/working/action_dets.json`.
